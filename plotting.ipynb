# Data Visualization with Matplotlib and Seaborn

This notebook demonstrates various data visualization techniques using Matplotlib and Seaborn. We'll cover line charts, bar charts, histograms, scatter plots, box plots, heatmaps, and advanced plotting techniques through practical exercises.

## Introduction

Data visualization is crucial for understanding patterns, trends, and relationships in data. This notebook covers:
- **Matplotlib**: A comprehensive library for creating static, animated, and interactive visualizations
- **Seaborn**: A statistical data visualization library built on matplotlib that provides a high-level interface for drawing attractive statistical graphics
- **Pandas**: Built-in plotting capabilities that integrate seamlessly with DataFrames

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

## Exercise 1: Line Chart with Population Growth

**Objective**: Plot a simple line chart using Matplotlib showing the population growth from 2000 to 2020.

**Formula**: Population growth = 2.5 + (1/10) * x, where x represents the year number: 1st, 2nd, ..., 20th.

**Requirements**: Create two lists - one for years and another for population values, then plot the line chart.

```python
years = list(range(2000, 2021))
print(years)
population = [2.5 + (1 / 10) * x for x in range(1, len(years) + 1)]
print(population)
plt.ylabel('Population')
plt.xlabel('Year')
plt.title('Population Growth')
plt.plot(years, population, 'green', marker='o', label='growth')
plt.legend(loc="lower right")
plt.show()
```

## Exercise 2: Revenue Normalization and Bar Charts

**Objective**: Create normalized revenue data and visualize it using bar charts and scatter plots.

**Data**: Products = ["A", "B", "C", "D", "E"] and Revenue = [500, 700, 1200, 900, 400]

**Task**: 
1. Create a normalized revenue list that scales values to the interval [0, 1]
2. Create bar charts and scatter plots for both original and normalized data

```python
Products = ["A", "B", "C", "D", "E"]
Revenue = [500, 700, 1200, 900, 400]
OldMin = min(Revenue)
OldMax = max(Revenue)
Normalized_Revenue = [(((OldValue - OldMin) * (1 - 0)) / (OldMax - OldMin)) + 0 for OldValue in Revenue]
print(Normalized_Revenue)
plt.ylabel('Revenue')
plt.xlabel('Products')
plt.title('Revenue per Product')
plt.bar(Products, Revenue)
plt.show()
plt.scatter(Products, Revenue)
plt.show()
plt.ylabel('Normalized Revenue')
plt.title('Normalized Revenue per Product')
plt.bar(Products, Normalized_Revenue)
plt.show()
plt.scatter(Products, Normalized_Revenue)
plt.show()
```

## Exercise 5: Department Salary Analysis with Box Plots

**Objective**: Analyze salary distribution across different departments using box plots.

**Requirements**:
1. Create departments list: ["HR", "IT", "Finance", "Marketing", "Sales"] * 5
2. Generate 25 random salary values between 40,000 and 100,000
3. Create a DataFrame and display box plots comparing salaries by department

```python
np.random.seed(0)
departments = ["HR", "IT", "Finance", "Marketing", "Sales"] * 5
salaries = np.random.randint(40_000, 100_001, 25)
df = pd.DataFrame({
    'Department': departments,
    'Salary': salaries
})
print(df)
res = df.groupby('Department')['Salary'].agg('sum')
print(res)
df.boxplot(column='Salary', by='Department')
plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Salary Distribution by Department")
plt.show()
```

## Exercise 6: Correlation Analysis with Heatmaps

**Objective**: Create and visualize correlations in a dataset using heatmaps.

**Requirements**:
1. Create a 10x10 square matrix M with random values between 16 and 36
2. Verify all values are in the correct range
3. Visualize correlations using a heatmap with 'coolwarm' colormap
4. Discuss the heatmap results

```python
np.random.seed(0)
M = np.random.randint(16, 37, (10, 10))
print(M)
in_range = lambda arr: all(16 <= val < 37 for val in [val for x in arr for val in x]) # arr.flatten()
result = "Values are in the right range" if in_range(M) else "Values are not in the right range"
print(result)
plt.imshow(M, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.title("Heatmap")
plt.show()
```

## Exercise 7: Advanced Population Growth Analysis

**Objective**: Plot population growth from 1980 to 2020 using a more complex formula.

**Formula**: Population = ((2.1 + (1/10) * x²) / (x² + 1)), where x represents year number: 1st, 2nd, ..., 40th.

**Requirements**:
1. Create lists for years (1980-2020) and population values
2. Print both lists
3. Plot the line chart

```python
x_axis = list(range(1980, 2021))
print(x_axis)
y_axis = [((2.1 + 0.1 * x ** 2) / (x ** 2 + 1)) for x in range(1, len(x_axis)+1)]
print(y_axis)
plt.xlabel('Year')
plt.ylabel('Population')
plt.plot(x_axis, y_axis)
plt.show()
```

## Exercise 10: Comprehensive Iris Dataset Analysis

**Objective**: Perform a complete visual analysis of the Iris dataset using various plot types.

**Analysis Tasks**:
1. Load dataset and display first five rows
2. Create scatter plots for sepal and petal measurements
3. Generate histograms for all measurement columns
4. Create box plots grouped by species
5. Generate pair plots comparing features by species
6. Create correlation heatmap
7. Display regression plots
8. Create count plots and KDE plots
9. Normalize and visualize species attributes

```python
# Load the Iris dataset
df = sns.load_dataset('iris')

print(df.head())
```

### Scatter Plot Analysis

Analyzing relationships between sepal and petal measurements.

```python
# 1. Scatter plot: SepalLengthCm vs SepalWidthCm
plt.figure(figsize=(10, 6))
sns.scatterplot(x='sepal_length', y='sepal_width', data=df)
plt.title('Scatter Plot: Sepal Length vs Sepal Width')
plt.show()
print("Strong correlation between x and y since there are two clusters and not 50,\
    and existence of outliers on the extremes")

# 2. Scatter plot: PetalLengthCm vs PetalWidthCm
plt.figure(figsize=(10, 6))
sns.scatterplot(x='petal_length', y='petal_width', data=df)
plt.title('Scatter Plot: Petal Length vs Petal Width')
plt.show()
print("Strong correlation between x and y since there are two clusters and not 50")
```

### Distribution Analysis

Examining the distribution of measurements using histograms.

```python
# 3. Histograms for all measurement columns
df.hist(figsize=(10,8), bins=20)
plt.suptitle("Histogram of Sepal and Petal Dimensions")
plt.show()
```

### Box Plot Analysis by Species

Comparing measurement distributions across different species.

```python
# 4. Box plot for SepalLengthCm by Species
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='sepal_length', data=df)
plt.title('Box Plot: Sepal Length by Species')
plt.show()

# 5. Box plot for PetalLengthCm by Species
plt.figure(figsize=(10, 6))
sns.boxplot(x='species', y='petal_length', data=df)
plt.title('Box Plot: Petal Length by Species')
plt.show()
```

### Comprehensive Relationship Analysis

Using pair plots to examine all feature relationships simultaneously.

```python
# 6. Pair plot
plt.figure(figsize=(12, 10))
pair_plot = sns.pairplot(df, hue='species')
plt.suptitle('Pair Plot of Iris Dataset Features by Species', y=1.02, fontsize=16)
plt.show()
```

### Correlation and Regression Analysis

Examining correlations and linear relationships between features.

```python
# 7. Correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap for Iris Dataset')
plt.show()

# 8. Regression plot: SepalLengthCm vs PetalLengthCm
plt.figure(figsize=(10, 6))
sns.regplot(x='petal_length', y='sepal_length', data=df)
plt.title('Regression Plot: Sepal Length vs Petal Length')
plt.show()

# 9. Regression plot: SepalWidthCm vs PetalWidthCm
plt.figure(figsize=(10, 6))
sns.regplot(x='petal_width', y='sepal_width', data=df, scatter_kws={'alpha':0.5})
plt.title('Regression Plot: Sepal Width vs Petal Width')
plt.show()
```

### Species Distribution and Density Analysis

Examining species distribution and measurement density.

```python
# 10. Count plot for Species
plt.figure(figsize=(10, 6))
sns.countplot(x='species', data=df)
plt.title('Count of Observations by Species')
plt.show()

# 11. KDE plot for SepalLengthCm
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='sepal_length', shade=True)
plt.title('KDE Plot: Distribution of Sepal Length by Species')
plt.show()
```
