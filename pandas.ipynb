# Pandas Data Analysis and Manipulation

This notebook covers the fundamental concepts of pandas, a powerful data analysis and manipulation library for Python. We'll explore Series, DataFrames, data filtering, sorting, grouping, and data merging operations.

## Introduction

Pandas is built on top of NumPy and provides high-performance, easy-to-use data structures and data analysis tools. The two primary data structures are:
- **Series**: One-dimensional labeled array
- **DataFrame**: Two-dimensional labeled data structure with columns of potentially different types

```python
import pandas as pd
```

## Series Operations

A Series is a one-dimensional array-like object containing a sequence of values and an associated array of data labels called its index.

### Creating Series

Series can be created from various data sources including lists, arrays, and dictionaries.

```python
ser = pd.Series(dtype="object")
print(ser)
data = pd.array(['h', 'e', 'l', 'l', 'o'])
ser = pd.Series(data)
print(ser)
```

## DataFrame Operations

A DataFrame is a 2-dimensional data structure that can store data of different types in columns.

### Creating DataFrames

DataFrames can be created from various sources including lists, dictionaries, and external files.

```python
df = pd.DataFrame()
print(df)
lst = ['Python', 'Software', 'is', 'Powerful', 'for', 'Data', "Analysis"]
df = pd.DataFrame(lst, columns=['Words'])
print(df)
```

### Loading Data from Files

Pandas can read data from various file formats. Here we load the famous Iris dataset.

```python
df = pd.read_csv("Datasets/Iris.csv")
df.head()
```

## Data Filtering and Selection

Pandas provides powerful methods for selecting and filtering data based on various criteria.

### Column Filtering

The `filter()` function is used to subset rows or columns of a DataFrame according to labels in the specified index. It filters based on labels, not content.

```python
df.filter(["Species", "SepalLengthCm"]).head()
```

## Data Sorting

Sorting is essential for data analysis and presentation.

### Sort Values Function

The `sort_values()` function sorts a DataFrame by one or more columns.

**Parameters:**
- `by`: Column name(s) to sort by (single or list)
- `axis`: 0 or 'index' for sorting rows, 1 or 'columns' for sorting columns
- `ascending`: Boolean, True for ascending, False for descending
- `inplace`: Boolean, if True, modifies the original DataFrame
- `kind`: Sorting algorithm: 'quicksort', 'mergesort' or 'heapsort'
- `na_position`: 'first' or 'last', defines position of NaN values

```python
df.sort_values(by='SepalLengthCm', ascending=True, inplace=False, kind='quicksort', na_position='last')
```

## GroupBy Operations

GroupBy operations involve splitting the data into groups, applying a function to each group, and combining the results.

### Understanding GroupBy

GroupBy mainly refers to a process involving one or more of the following steps:
- **Splitting**: Split data into groups by applying some conditions on datasets
- **Applying**: Apply a function to each group independently
- **Combining**: Combine different datasets after applying groupby

```python
data1 = {
    'Name': ['Jai', 'Anuj', 'Jai', 'Princi', 'Gaurav', 'Anuj', 'Princi', 'Abhi'], 
    'Age': [27, 24, 22,32, 33, 36, 27, 32], 
    'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj', 'Jaunpur', 'Kanpur', 
                'Allahabad', 'Aligarh'], 
    'Qualification': ['Msc', 'MA', 'MCA', 'Phd', 'B.Tech', 'B.com', 'Msc', 'MA']
}
df = pd.DataFrame(data1)
print("Original DataFrame")
print(df)
gk = df.groupby('Name')
print("After creating Groups")
gk.first()
```

### Aggregation Operations

Aggregation is a process where we compute summary statistics about each group, such as group sums or means.

```python
result = gk['Age'].aggregate('sum')
print(result)
```

### Transformation Operations

Transformation performs group-specific computations and returns a like-indexed result. For example, filling NAs within groups with a value derived from each group.

### Filtration Operations

Filtration is a process where we discard some groups according to a group-wise computation that evaluates True or False, such as filtering out data based on the group sum or mean.

## Data Combining Operations

Pandas provides several methods to combine DataFrames including concatenation, merging, and joining.

### Concatenation

Concatenation combines DataFrames along a particular axis (rows or columns).

```python
data1 = {
    'key': ['K0', 'K1', 'K2', 'K3'],
    'Name': ['Jai', 'Princi', 'Gaurav', 'Anuj'],
    'Age': [27, 24, 22, 32]
}
data2 = {
    'key': ['K0', 'K1', 'K2', 'K3'],
    'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
    'Qualification': ['Btech', 'B.A', 'Bcom', 'B.hons']
}
df = pd.DataFrame(data1)
df1 = pd.DataFrame(data2)
display(df, df1)
res = pd.concat([df, df1], axis=1)
print(res)
# axis=1: Concatenates along the columns (adding columns side by side, horizontally).
```

### Merge Operations

Merge combines DataFrames based on common columns or indices, similar to SQL joins.

```python
res = pd.merge(df, df1, on='key')
print(res)
```

### Join Operations

Join is used for combining the columns of two potentially differently indexed DataFrames into a single result DataFrame.

**Syntax:** `separator.join(iterable)`
- `separator`: The string placed between elements of the iterable
- `iterable`: A sequence of strings (e.g., list, tuple, etc.) to join together

```python
data1 = {
    'Name': ['Jai', 'Princi', 'Gaurav', 'Anuj'],
    'Age': [27, 24, 22, 32]
}
data2 = {
    'Address': ['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
    'Qualification': ['Btech', 'B.A', 'Bcom', 'B.hons']
}
df = pd.DataFrame(data1, index=['K0', 'K1', 'K2', 'K3'])
df1 = pd.DataFrame(data2, index=['K0', 'K2', 'K3', 'K4'])
display(df, df1)
res = df.join(df1)
print(res)
```
