# Seaborn Data Visualization

This notebook demonstrates advanced data visualization techniques using Seaborn, focusing on pairplots and their various customization options. Seaborn provides a high-level interface for drawing attractive and informative statistical graphics.

## Introduction

Seaborn is a statistical data visualization library built on matplotlib. It provides:
- Beautiful default styles and color palettes
- Functions for visualizing univariate and bivariate distributions
- Tools for fitting and visualizing linear regression models
- Built-in themes for styling matplotlib graphics
- High-level interface for drawing attractive statistical graphics

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Seaborn Pairplot

The pairplot is one of the most useful functions in Seaborn for exploratory data analysis. It creates a grid of scatter plots for each pair of variables in a dataset.

### Pairplot Syntax and Parameters

**Syntax:** `seaborn.pairplot(data, **kwargs)`

**Key Parameters:**
- `data`: Tidy (long-form) dataframe where each column is a variable and each row is an observation
- `hue`: Variable in "data" to map plot aspects to different colors
- `palette`: Dict or seaborn color palette for color mapping
- `{x, y}`: Lists of variable names (optional)
- `dropna`: Boolean, optional parameter for handling missing values

### Basic Pairplot Example

Let's start with a basic pairplot using the built-in tips dataset.

```python
df = sns.load_dataset('tips')
print(df.head())
sns.pairplot(df)
plt.show()
```

## Customizing Pairplots

Pairplots can be customized in various ways to focus on specific variables or highlight patterns.

### Plotting Selected Variables

You can specify which variables to include in the pairplot using the `vars` parameter.

```python
selected_vars = ['total_bill', 'tip']
sns.pairplot(df, vars=selected_vars)
plt.show()
```

### Using the 'hue' Parameter

The `hue` parameter allows you to color-code data points based on a categorical variable.

```python
sns.pairplot(df, vars=selected_vars, hue='size')
plt.show()
```

### Using the 'palette' Parameter

The `palette` parameter is always used with `hue` to specify custom color schemes.

```python
sns.pairplot(df, vars=selected_vars, hue='size', palette='husl')
plt.show()
```

## Diagonal Plot Customization

The diagonal of a pairplot shows the distribution of individual variables.

### Using 'diag_kind' Parameter

The `diag_kind` parameter specifies the type of plot to display along the diagonal axis, representing the univariate distribution of each variable.

**Options include:**
- `'hist'` for histograms
- `'kde'` for kernel density estimates
- `'scatter'` for scatter plots

Choose based on the nature of the data and analysis goals.

```python
sns.pairplot(df, diag_kind='kde')
plt.show()
```

## Off-Diagonal Plot Customization

The off-diagonal plots show relationships between different variables.

### Using 'kind' Parameter

The `kind` parameter specifies the type of plot for the off-diagonal elements.

```python
sns.pairplot(df, kind='reg')
plt.show()
```

## Advanced Customization

Pairplots offer additional customization options for enhanced visualization.

### Using 'markers' Parameter

The `markers` parameter allows you to specify different markers for different categories when using the `hue` parameter.

```python
sns.pairplot(df, hue='sex', markers=["o", "s"])
plt.show()
# This function is great for exploring relationships between multiple numerical variables 
# in your dataset while visually encoding categorical data.
```

## Advanced Customization with FaceGrid

For more advanced customization, you can access the underlying FaceGrid object and modify it further.

```python
g = sns.pairplot(df, hue='day')
g.fig.suptitle("Pairplot of Tips Dataset", y=1.02) # Add title
g.set(xticks=[], yticks=[]) # Remove tick labels
plt.show()
```
